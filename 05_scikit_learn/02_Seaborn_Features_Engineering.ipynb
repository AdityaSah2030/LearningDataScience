Handling Imbalance Data set

```python
np.random.seed(3)
n_samples=1000
class_0_ratio=0.9
n_class_0=int(n_samples*class_0_ratio)
n_class_1=n_samples-n_class_0
n_class_0,n_class_1
```

```python
df1=pd.DataFrame({

      "Column-1":np.random.normal(loc=0,scale=1,size=n_class_0),
      "Column-2":np.random.normal(loc=0,scale=1,size=n_class_0),
      "Target":[0]*n_class_0
})
df1



```

```python
df2=pd.DataFrame({

      "Column-1":np.random.normal(loc=2,scale=1,size=n_class_1),
      "Column-2":np.random.normal(loc=2,scale=1,size=n_class_1),
      "Target":[1]*n_class_1

})
df2
```

```python
df_merge=pd.concat([df1,df2]).reset_index(drop=True)
df_merge
```

```python
#Seperate the major class and minor
df_major=df_merge[df_merge['Target']==0]
df_minor=df_merge[df_merge['Target']==1]
df_major.shape,df_minor.shape
```

```python
#Start the sampling operation
from sklearn.utils import resample
```

```python
#Upsampling
df_minor_upsample=resample(df_minor,replace=True,n_samples=len(df_major),random_state=42)
df_minor_upsample.shape
```

```python
df_upsample_final=pd.concat([df_minor_upsample,df_major])
df_upsample_final['Target'].value_counts()
```

SMOTE

```python
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=1000,n_features=2,n_redundant=0,n_clusters_per_class=1,weights=[0.90],random_state=12)
X
```

```python
y
```

```python
df1=pd.DataFrame(X,columns=["Column-1","Column-2"])
df1
```

```python
df2=pd.DataFrame(y,columns=["Target"])
df2
```

```python
final_df=pd.concat([df1,df2],axis=1)
final_df.head(10)
```

```python
final_df['Target'].value_counts()
```

```python
plt.scatter(x=final_df['Column-1'],y=final_df['Column-2'],c=final_df['Target'])
```
